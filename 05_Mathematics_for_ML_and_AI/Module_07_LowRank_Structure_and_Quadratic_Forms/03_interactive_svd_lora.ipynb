# Interactive Lab: Low-Rank Approximations and SVD
Explore how low-rank truncation compresses data while retaining maximum variance.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate a synthetic low-rank matrix with noise
np.random.seed(42)
true_u = np.random.randn(50, 3)
true_v = np.random.randn(3, 40)
clean_mat = true_u @ true_v
noise = 0.1 * np.random.randn(50, 40)
data_mat = clean_mat + noise

# SVD decomposition
U, S, Vt = np.linalg.svd(data_mat, full_matrices=False)

plt.figure(figsize=(8, 4))
plt.plot(np.arange(1, len(S) + 1), S, 'bo-')
plt.title('Singular Value Spectrum (Scree Plot)')
plt.xlabel('Component index')
plt.ylabel('Singular value')
plt.grid(True)
plt.show()
print(f'Top 3 singular values hold {np.sum(S[:3]**2) / np.sum(S**2) * 100:.1f}% of total energy!')
```
