# 🎨 Interactive Lab: Eigenvalues & Space Stretch

Observe how eigenvectors maintain their direction under matrix transformation.

```python
import numpy as np

A = np.array([[3.0, 1.0], [0.0, 2.0]])
eigenvalues, eigenvectors = np.linalg.eig(A)

print('Eigenvalues:', eigenvalues)
print('Eigenvectors:\n', eigenvectors)
```
