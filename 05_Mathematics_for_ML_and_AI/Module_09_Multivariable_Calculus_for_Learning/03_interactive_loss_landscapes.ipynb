# Interactive Lab: Loss Landscapes and Gradients
Visualize 2D contour optimization surfaces and gradient descent trajectories.

```python
import matplotlib.pyplot as plt
import numpy as np

# 2D Bowl Loss surface
x = np.linspace(-4, 4, 100)
y = np.linspace(-4, 4, 100)
X, Y = np.meshgrid(x, y)
Z = 0.5 * X**2 + 2.0 * Y**2

plt.figure(figsize=(6, 5))
cp = plt.contour(X, Y, Z, levels=20, cmap='viridis')
plt.clabel(cp, inline=True, fontsize=8)
plt.plot(0, 0, 'r*', markersize=12, label='Global Minimum')
plt.title('Loss Contours: Steeper along Y than X')
plt.xlabel('x1')
plt.ylabel('x2')
plt.legend()
plt.grid(True)
plt.show()
```
