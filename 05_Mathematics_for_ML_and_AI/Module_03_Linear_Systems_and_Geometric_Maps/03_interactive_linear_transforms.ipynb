# 🎨 Interactive Lab: Linear Transformations in 2D Space

Watch how matrices bend, stretch, and rotate 2D coordinates in real-time.

```python
import numpy as np

# Define a 2D shear/rotation transformation matrix
A = np.array([
    [2.0, 1.0],
    [0.5, 1.5]
])

# Standard basis vectors
i_hat = np.array([1.0, 0.0])
j_hat = np.array([0.0, 1.0])

print('Transformed i_hat:', A @ i_hat)
print('Transformed j_hat:', A @ j_hat)
print('Determinant (Area Scaling Factor):', np.linalg.det(A))
```
