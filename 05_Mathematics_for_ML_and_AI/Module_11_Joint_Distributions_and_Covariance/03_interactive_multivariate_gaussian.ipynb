# Interactive Lab: Multivariate Gaussians & Covariance Contours
Visualize how covariance tilts the probability density ellipse.

```python
import matplotlib.pyplot as plt
import numpy as np

# Grid for 2D density
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
pos = np.dstack((X, Y))

# Correlated covariance: sigma_x=1, sigma_y=1, cov_xy=0.8
mu = np.array([0.0, 0.0])
cov = np.array([[1.0, 0.8], [0.8, 1.0]])
inv_cov = np.linalg.inv(cov)

diff = pos - mu
quad = np.einsum('...k,kl,...l->...', diff, inv_cov, diff)
density = np.exp(-0.5 * quad) / (2 * np.pi * np.sqrt(np.linalg.det(cov)))

plt.figure(figsize=(6, 5))
plt.contour(X, Y, density, levels=10, cmap='magma')
plt.title('Tilted 2D Gaussian Contour (cov=0.8)')
plt.xlabel('X1')
plt.ylabel('X2')
plt.grid(True)
plt.show()
```
