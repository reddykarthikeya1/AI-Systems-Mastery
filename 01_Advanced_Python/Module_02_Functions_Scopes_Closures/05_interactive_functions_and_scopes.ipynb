## 1. Parameter Types: `/`, `*`, `*args`, and `**kwargs`

Run this cell to see how Python enforces positional-only and keyword-only constraints.

```python
def configure_server(host, port, /, *, debug=False, log_level="INFO"):
    print(f"Starting server on {host}:{port} [Debug={debug}, Log={log_level}]")

# Correct usage:
configure_server("127.0.0.1", 8000, debug=True, log_level="DEBUG")

# Try uncommenting this to see the TypeError:
# configure_server(host="127.0.0.1", port=8000, debug=True)
```

## 2. Argument Unpacking with `*` and `**`

Unpack lists, tuples, and dictionaries directly into function calls.

```python
def calculate_total(price, tax_rate, discount=0.0):
    subtotal = price - discount
    return subtotal * (1 + tax_rate)

# Unpacking a list/tuple (*args)
order_data = (100.0, 0.08)
print(f"Total from tuple: ${calculate_total(*order_data):.2f}")

# Unpacking a dictionary (**kwargs)
order_dict = {"price": 200.0, "tax_rate": 0.08, "discount": 25.0}
print(f"Total from dict : ${calculate_total(**order_dict):.2f}")
```

## 3. Variable Scope (LEGB) & Closures

A **closure** is a function that retains access to variables in its enclosing scope even after the outer function has returned.

```python
def create_hero(name: str, starting_hp: int = 100):
    hp = starting_hp  # Enclosing variable inside factory

    def take_damage(amount: int) -> int:
        nonlocal hp
        hp = max(0, hp - amount)
        print(f"{name} took {amount} damage! Current HP: {hp}")
        return hp

    def heal(amount: int) -> int:
        nonlocal hp
        hp = min(starting_hp, hp + amount)
        print(f"{name} healed for {amount}! Current HP: {hp}")
        return hp

    return take_damage, heal

damage_warrior, heal_warrior = create_hero("Conan the Barbarian", 120)
damage_warrior(30)
damage_warrior(50)
heal_warrior(20)
```

### Inspecting the Closure Memory Backpack (`__closure__`)
Let's peek inside the function's memory cell to see the stored value.

```python
print("Closure cells:", damage_warrior.__closure__)
print("Stored HP value in memory cell:", damage_warrior.__closure__[0].cell_contents)
```

## 4. First-Class Functions & Lambdas

Functions can be passed as arguments and sorted dynamically.

```python
monsters = [
    {"name": "Goblin", "threat_level": 2, "loot_gold": 15},
    {"name": "Dragon", "threat_level": 10, "loot_gold": 5000},
    {"name": "Orc Chieftain", "threat_level": 6, "loot_gold": 250},
]

# Sort by threat level ascending
sorted_by_threat = sorted(monsters, key=lambda m: m["threat_level"])
print("Sorted by Threat Level:")
for m in sorted_by_threat:
    print(f"  [{m['threat_level']:02d}] {m['name']} (Gold: {m['loot_gold']})")
```

## 5. Standard Library: Random Rolls, Datetime & Timezones

```python
import random
from datetime import UTC, datetime
from zoneinfo import ZoneInfo

# Random dice rolls
d20_roll = random.randint(1, 20)
print(f"Rolling D20... Result: {d20_roll} {'(CRITICAL HIT! 💥)' if d20_roll == 20 else ''}")

# Modern timezone handling with datetime & zoneinfo
utc_now = datetime.now(UTC)
print(f"Current Time (UTC): {utc_now.strftime('%Y-%m-%d %H:%M:%S %Z')}")

tokyo_time = utc_now.astimezone(ZoneInfo("Asia/Tokyo"))
print(f"Current Time (Tokyo): {tokyo_time.strftime('%Y-%m-%d %H:%M:%S %Z')}")
```
