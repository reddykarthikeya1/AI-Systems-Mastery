```python
# Step 2: Testing Pydantic Telemetry Schema Validation
from pydantic import BaseModel, Field


class DeviceTelemetry(BaseModel):
    device_id: str
    temperature: float = Field(ge=-40.0, le=120.0)
    pressure: float = Field(gt=0.0)

event = DeviceTelemetry(device_id='turbine_alpha_4', temperature=68.2, pressure=101.3)
print(f'Validated Ingress Event: {event.model_dump()}')
```

```python
# Step 3: Simulating Ingestion Pipeline into Stream Buffer
buffer = []
def enqueue_telemetry(payload: dict):
    buffer.append(payload)
    return len(buffer)

enqueue_telemetry(event.model_dump())
print(f'Ingress buffer size: {len(buffer)} events')
```
