## 1. Variables, Data Types & Memory Addresses (`id()`)

In Python, variables are sticky name tags attached to objects stored in RAM. You can inspect an object's memory address using `id()` and its type using `type()`.

```python
x = 100
print(f"Value of x : {x}")
print(f"Type of x  : {type(x).__name__}")
print(f"Memory ID  : {id(x)}")

# Reassign x to a string (Dynamic typing in action)
x = "Python Mastery"
print("\nAfter reassignment:")
print(f"Value of x : {x}")
print(f"Type of x  : {type(x).__name__}")
print(f"New Memory ID: {id(x)}")
```

## 2. The Floating-Point Surprise: Why $0.1 + 0.2 \neq 0.3$

Run this cell to see binary floating-point representation in action!

```python
import math
from decimal import Decimal

a = 0.1
b = 0.2
c = a + b

print(f"Raw sum of 0.1 + 0.2       : {c:.20f}")
print(f"Does 0.1 + 0.2 == 0.3?     : {c == 0.3}")
print(f"Using math.isclose(c, 0.3) : {math.isclose(c, 0.3)}")

# Exact financial calculation using Decimal:
d1 = Decimal("0.1")
d2 = Decimal("0.2")
print(f"Decimal('0.1') + Decimal('0.2') == Decimal('0.3'): {d1 + d2 == Decimal('0.3')}")
```

## 3. Identity (`is`) vs. Equality (`==`)

- `==` checks if two values **look the same** (value equality).
- `is` checks if two variables **are the exact same object in RAM** (pointer identity).

```python
list_a = [1, 2, 3]
list_b = [1, 2, 3]

print(f"list_a == list_b : {list_a == list_b}  (Same values inside)")
print(f"list_a is list_b : {list_a is list_b}  (Separate memory allocations)")
print(f"id(list_a) = {id(list_a)}, id(list_b) = {id(list_b)}")

list_c = list_a
print(f"\nlist_a is list_c : {list_a is list_c}  (Both point to the same memory)")
```

## 4. String Indexing & Slicing Magic

Syntax: `string[start:stop:step]` (Remember: `stop` is exclusive!).

```python
message = "AdvancedPython"

print(f"Original string     : {message}")
print(f"First character [0] : {message[0]}")
print(f"Last character [-1] : {message[-1]}")
print(f"Slice [0:8]         : {message[0:8]}")
print(f"Slice [8:]          : {message[8:]}")
print(f"Every 2nd char [::2]: {message[::2]}")
print(f"Reversed string [::-1]: {message[::-1]}")
```

## 5. Modern f-String Formatting

Format currency, percentages, decimals, and tables easily.

```python
principal = 250000.50
annual_interest = 0.0675
years = 30

print("=== Financial Summary ===")
print(f"Principal Loan   : ${principal:,.2f}")
print(f"Interest Rate    : {annual_interest:.2%}")
print(f"Duration         : {years} years")
```

## 6. Modern Pattern Matching (`match - case`)

Python 3.10+ provides expressive structural pattern matching.

```python
def handle_http_status(status_code: int) -> str:
    match status_code:
        case 200 | 201:
            return "Success (OK / Created)"
        case 400:
            return "Client Error: Bad Request"
        case 401 | 403:
            return "Security Error: Unauthorized / Forbidden"
        case 404:
            return "Not Found"
        case 500 | 502 | 503:
            return "Server Error"
        case _:
            return f"Unknown status: {status_code}"

print(handle_http_status(200))
print(handle_http_status(404))
print(handle_http_status(503))
print(handle_http_status(999))
```

## 7. Loops: `for-else` Prime Number Search

The loop `else` block executes **only when the loop finishes without encountering a `break`**.

```python
def check_prime(number: int) -> None:
    if number < 2:
        print(f"{number} is not prime.")
        return

    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            print(f"{number} is composite (divisible by {divisor}).")
            break
    else:
        print(f"{number} is a PRIME number! 🌟")

check_prime(29)
check_prime(49)
check_prime(1)
```
