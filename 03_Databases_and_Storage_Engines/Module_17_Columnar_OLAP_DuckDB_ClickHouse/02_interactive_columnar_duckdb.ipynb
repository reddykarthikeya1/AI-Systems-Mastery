## 1. Engine Initialization & Setup

Importing the module's Track A internal simulation engine and instantiating state.


```python
from columnar_engine import DictionaryEncoder, RLEEncoder

# Run-Length Encoding (RLE) on sorted categorical columns
status_column = ["ACTIVE"] * 100 + ["PENDING"] * 20 + ["CANCELLED"] * 5
rle_encoded = RLEEncoder.encode(status_column)
rle_decoded = RLEEncoder.decode(rle_encoded)

print(f"Original items: {len(status_column)} -> RLE Compressed runs: {len(rle_encoded)}")
print("RLE Runs:", rle_encoded)
```

## 2. Core Architectural Operations & State Mutation

Executing data mutations, transactions, or indexing procedures.


```python
# Dictionary Encoding for high-cardinality string columns
regions = ["US-EAST", "EU-WEST", "US-EAST", "AP-SOUTH", "US-EAST"] * 20
dictionary, codes = DictionaryEncoder.encode(regions)
print(f"Dictionary Table: {dictionary}")
print(f"Integer Code Stream (first 10): {codes[:10]}")

decoded_regions = DictionaryEncoder.decode(dictionary, codes)
print("Dictionary decoded matches original:", decoded_regions == regions)
```

## 3. Performance Micro-Benchmarking & Invariant Verification

Evaluating execution latency, cache hits, or computational trade-offs.


```python
# Verify Columnar Invariants
assert rle_decoded == status_column
assert decoded_regions == regions
print("[+] Columnar RLE and Dictionary Encoding invariants verified successfully!")
```
