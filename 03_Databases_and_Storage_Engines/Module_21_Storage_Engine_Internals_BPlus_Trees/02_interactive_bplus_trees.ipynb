## 1. Engine Initialization & Setup

Importing the module's Track A internal simulation engine and instantiating state.


```python
from storage_engine import BufferPoolManager, SlottedPage, DiskManager

# Initialize 1024-Byte Slotted Page
page = SlottedPage(page_id=1, page_size=1024)

# Insert variable-length binary records
# Record pointers grow forward from slot array; data grows backwards from page bottom!
s0 = page.insert_tuple(b"Customer_Record_Alice")
s1 = page.insert_tuple(b"Customer_Record_Bob_With_Longer_Payload")

print(f"Slotted Page Free Space: {page.free_space()} bytes (Tuple count: {page.tuple_count})")
print(f"Slot 0 read: {page.get_tuple(s0).decode()}")
print(f"Slot 1 read: {page.get_tuple(s1).decode()}")
```

## 2. Core Architectural Operations & State Mutation

Executing data mutations, transactions, or indexing procedures.


```python
# Buffer Pool Manager Clock Sweep Eviction Simulation
disk_mgr = DiskManager()
p0 = disk_mgr.allocate_page()
p1 = disk_mgr.allocate_page()
p2 = disk_mgr.allocate_page()

bpm = BufferPoolManager(pool_size=3, disk_manager=disk_mgr)
bpm.fetch_page(p0)
bpm.fetch_page(p1)
bpm.fetch_page(p2)

print("Buffer Pool filled to capacity (3 pages).")
```
