## 1. Engine Initialization & Setup

Importing the module's Track A internal simulation engine and instantiating state.


```python
from vector_engine import FlatVectorIndex, VectorItem

# Initialize Flat Exact kNN Vector Index
v_index = FlatVectorIndex()

# Add embeddings
v_index.add(VectorItem("vec_1", [0.1, 0.2, 0.9, 0.0], {"topic": "ai"})) # Tech document
v_index.add(VectorItem("vec_2", [0.8, 0.7, 0.1, 0.1], {"topic": "finance"})) # Finance document
v_index.add(VectorItem("vec_3", [0.15, 0.25, 0.85, 0.05], {"topic": "ai"})) # Tech related document
print("Indexed 3 dense 4-dimensional vectors.")
```

## 2. Core Architectural Operations & State Mutation

Executing data mutations, transactions, or indexing procedures.


```python
# Cosine Distance kNN Query (lower distance = higher similarity)
query_vec = [0.12, 0.22, 0.88, 0.02]
neighbors = v_index.search(query_vec, top_k=2)

print("Nearest Neighbors:")
for item, dist in neighbors:
    print(f"  ID: {item.item_id:<8} Topic: {item.payload['topic']} Cosine Distance: {dist:.4f}")
```

## 3. Performance Micro-Benchmarking & Invariant Verification

Evaluating execution latency, cache hits, or computational trade-offs.


```python
# Verify Vector Invariants
assert neighbors[0][0].item_id in ("vec_1", "vec_3"), "Query vector must match closest semantic vector"
assert neighbors[0][1] < 0.05, "Distance to near-identical vector must be near 0.0"
print("[+] Vector Database kNN Similarity invariants verified successfully!")
```
