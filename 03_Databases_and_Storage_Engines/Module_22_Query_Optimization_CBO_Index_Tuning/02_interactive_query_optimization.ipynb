## 1. Engine Initialization & Setup

Importing the module's Track A internal simulation engine and instantiating state.


```python
from query_optimizer import CostEstimator, PhysicalJoins, TableStatistics

# Table Statistics for Cost Estimator
stats = TableStatistics(
    table_name="orders",
    total_pages=1000,
    total_tuples=100000,
    distinct_counts={"customer_id": 50000, "status": 2},
    mcv_frequencies={"status": {"COMPLETED": 0.90, "CANCELLED": 0.10}}
)
cbo = CostEstimator()

# Compare Cost of Full Sequential Table Scan vs B-Tree Index Scan
cost_seq = cbo.estimate_seq_scan_cost(stats)
cost_idx = cbo.estimate_index_scan_cost(stats, selectivity=0.001) # 0.1% selectivity

print(f"Cost of Full Sequential Scan: {cost_seq:.2f} cost units")
print(f"Cost of Index Scan (0.1% selectivity): {cost_idx:.2f} cost units")
print(f"Optimal Choice: {'Index Scan' if cost_idx < cost_seq else 'Seq Scan'}")
```

## 2. Core Architectural Operations & State Mutation

Executing data mutations, transactions, or indexing procedures.


```python
# Physical Join Algorithms: Hash Join
outer_table = [{"id": i, "val": f"v_{i}"} for i in range(10)]
inner_table = [{"id": i, "extra": f"e_{i}"} for i in range(5, 15)]

hash_join_res = PhysicalJoins.hash_join(outer_table, inner_table, "id", "id")
print(f"Hash Join matches found: {len(hash_join_res)}")
```

## 3. Performance Micro-Benchmarking & Invariant Verification

Evaluating execution latency, cache hits, or computational trade-offs.


```python
# Verify CBO Invariants
assert cost_idx < cost_seq, "Index scan must be cheaper for 0.1% selectivity"
assert len(hash_join_res) == 5, "Keys 5..9 match between outer and inner tables"
print("[+] Cost-Based Optimizer (CBO) and Join Selection invariants verified successfully!")
```
